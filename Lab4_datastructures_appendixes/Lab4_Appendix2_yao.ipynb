```python
import random
import pandas as pd
import numpy as np

LAST_NAME = "Yao"
STUDENT_ID = "TUPM-26-1186"
SEED_NUM = int(STUDENT_ID[-1]) #this is gonna be 6 again idk why i keep doing the full thing when i can just do this but whatever
FAVORITE_ARTIST = "BEN AND BEN"#I just imported these from first appendix cause they all use the same inputs nmn

random.seed(SEED_NUM + len(FAVORITE_ARTIST))

print(f"Inputs Loaded Successfully for: {LAST_NAME}")
print(f"Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
```

```python
student_names = [f"{LAST_NAME}_Student_{chr(65+i)}" for i in range(5)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
raw_records = []
for student in student_names:
    for subject in subjects:
        base_score = int((SEED_NUM + sum(ord(c) for c in FAVORITE_ARTIST) * random.randint(1, 3)) % 51) + 50
        score = min(max(base_score, 50), 100)
        raw_records.append({
            "Student_Name": student,
            "Subject": subject,
            "Score": score
            })


raw_records.append(raw_records[0].copy())
raw_records.append({
    "Student_Name": student_names[0],
    "Subject": "PROGRAMMING",
    "Score": np.nan
})

df_perf_raw = pd.DataFrame(raw_records)
print(f"Total Raw Records Generated: {len(df_perf_raw)}")
display(df_perf_raw.head(10))
```

```python
df_clean_org = df_perf_raw.drop_duplicates().copy()

student_pivot = df_clean_org.pivot_table(index="Student_Name", columns="Subject", values="Score")
display(student_pivot)

#I like how Gemini handles the usage of cells, they have a raw data cell and then a clean up cell its really cool
```

```python
def get_performance_category(score):
    if pd.isna(score):
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Above Average"
    elif score >= 75:
        return "Average"
    else:
        return "Passing"

df_perf_raw["Performance_Category"] = df_perf_raw["Score"].apply(get_performance_category)
repeated_records = df_perf_raw[df_perf_raw.duplicated(keep=False)]
print("--- Repeated Records Found ---")
display(repeated_records)
incomplete_records = df_perf_raw[df_perf_raw["Score"].isna()]
print("\n--- Incomplete Records Found ---")
display(incomplete_records)
```

```python
print("--- Summary by Student ---")
student_summary = df_perf_raw.groupby("Student_Name")["Score"].agg(
    Total_Subjects_Taken="count",
    Average_Score="mean",
    Highest_Score="max",
    Lowest_Score="min"
).reset_index()
display(student_summary)

print("\n--- Summary by Subject ---")
subject_summary = df_perf_raw.groupby("Subject")["Score"].agg(
    Total_Students="count",
    Average_Score="mean",
    Highest_Score="max",
    Lowest_Score="min"
).reset_index()
display(subject_summary)
```

```python
ordered_performance = df_perf_raw.sort_values(by=["Student_Name", "Score"], ascending=[True, False]).reset_index(drop=True)

print("==================================================")
print(" FINAL CONSOLIDATED STUDENT PERFORMANCE REPORT")
print("==================================================")
display(ordered_performance)
```
