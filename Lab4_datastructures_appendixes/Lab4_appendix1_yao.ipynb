```python
#song lyrics included in the code cause why not, also this is the cell meant for the data inputs
#THIS IS ME
LAST_NAME = "Yao"
STUDENT_ID = "TUPM-26-1186"
SEED_NUM = int(STUDENT_ID[-1]) #this is gonna be 6 again idk why i keep doing the full thing when i can just do this but whatever
FAVORITE_ARTIST = "BEN AND BEN"
#LOOK OUT CAUSE HERE I COME
import random
import pandas as pd
random.seed(SEED_NUM + len(FAVORITE_ARTIST))
#AND THEYRE MARCHING ONTO THE BEAT I DRUMMM
print(f"Inputs Loaded Successfully for: {LAST_NAME}")
print(f"Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")

```

```python
#helped with ai kasi we're all cramming studying for midterms habang ginagawa namin po ito sa library

sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

raw_records = []

for source in sources:
    for cat in categories:
        base_val = round(random.uniform(10.0, 500.0) + sum(ord(c) for c in FAVORITE_ARTIST) % 50, 2)
        raw_records.append({"Test_Source": source, "Category": cat, "Measured_Value": base_val})

raw_records.append(raw_records[0].copy())
raw_records.append({"Test_Source": sources[0], "Category": "TEMPERATURE", "Measured_Value": 99.99})
raw_records.append({"Test_Source": sources[1], "Category": "VOLTAGE", "Measured_Value": 220.50})
df_raw = pd.DataFrame(raw_records)
print(f"Total Raw Records Generated: {len(df_raw)}")
display(df_raw.head(10))

print (display) #putitng this here for me to see it, also posterity
```

```python
#Organizing cell, thanks Gemini!!


organized_df = df_raw.groupby(["Test_Source", "Category"])["Measured_Value"].agg(['count', 'mean', 'min', 'max']).reset_index()
organized_df.rename(columns={'count': 'Record_Count', 'mean': 'Average_Value', 'min': 'Min_Value', 'max': 'Max_Value'}, inplace=True)

display(organized_df)
```

```python
duplicates = df_raw[df_raw.duplicated(keep=False)]
print("--- Repeated Records Found ---")
display(duplicates)

distinct_sources = df_raw["Test_Source"].unique()
distinct_categories = df_raw["Category"].unique()
print(f"\nDistinct Test Sources ({len(distinct_sources)}): {list(distinct_sources)}")
print(f"Distinct Categories ({len(distinct_categories)}): {list(distinct_categories)}")
```

```python
categorical_summary = df_raw.groupby("Category")["Measured_Value"].agg(
    Total_Entries="count",
    Mean_Value="mean",
    Median_Value="median",
    Std_Dev="std",
    Max_Value="max",
    Min_Value="min"
).reset_index()

display(categorical_summary)
```

```python
ordered_results = df_raw.sort_values(by=["Category", "Measured_Value"], ascending=[True, False]).reset_index(drop=True)

display(ordered_results.head(15))
```

```python
final_dataset = df_raw.drop_duplicates().reset_index(drop=True)

print("==========================================")
print(" FINAL ORGANIZED & SUMMARIZED TEST DATASET")
print("==========================================")
print(f"Total Clean Records: {len(final_dataset)}")
display(final_dataset)
```
