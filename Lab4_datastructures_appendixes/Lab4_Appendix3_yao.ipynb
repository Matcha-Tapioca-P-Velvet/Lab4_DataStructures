```python
import random
import pandas as pd

LAST_NAME = "Yao"
STUDENT_ID = "TUPM-26-1186"
SEED_NUM = int(STUDENT_ID[-1]) #this is gonna be 6 again idk why i keep doing the full thing when i can just do this but whatever
FAVORITE_ARTIST = "BEN AND BEN"#I just imported these from first appendix cause they all use the same inputs nmn

random.seed(SEED_NUM + len(FAVORITE_ARTIST))

print(f"Inputs Loaded Successfully for: {LAST_NAME}")
print(f"Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
```

```python
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products_prices = {
    "NOTEBOOK": 50.00,
    "PEN": 15.50,
    "USB": 250.00,
    "HEADPHONES": 750.00,
    "CALCULATOR": 1200.00
}

raw_transactions = []

for customer in customers:
    for product in products_prices.keys():
        qty = int((SEED_NUM + sum(ord(c) for c in FAVORITE_ARTIST)) % 5) + 1
        raw_transactions.append({
            "Customer": customer,
            "Product": product,
            "Quantity": qty,
            "Unit_Price": products_prices[product]
        })
raw_transactions.append(raw_transactions[0].copy())

df_trans_raw = pd.DataFrame(raw_transactions)
print(f"Total Raw Transactions Generated: {len(df_trans_raw)}")
display(df_trans_raw.head(10))

```

```python
df_trans_raw["Total_Value"] = df_trans_raw["Quantity"] * df_trans_raw["Unit_Price"]

df_organized = df_trans_raw[["Customer", "Product", "Quantity", "Unit_Price", "Total_Value"]].copy()

total_dataset_value = df_organized["Total_Value"].sum()
print(f"Overall Dataset Transaction Value: ${total_dataset_value:,.2f}\n")
display(df_organized.head(10))
```

```python
repeated_purchases = df_organized[df_organized.duplicated(keep=False)]
print("--- Repeated Purchases Found ---")
display(repeated_purchases)

distinct_customers = df_organized["Customer"].unique()
distinct_products = df_organized["Product"].unique()

print(f"\nDistinct Customers ({len(distinct_customers)}): {list(distinct_customers)}")
print(f"Distinct Products ({len(distinct_products)}): {list(distinct_products)}")
```

```python
print("--- Customer Summaries ---")
customer_summary = df_organized.groupby("Customer").agg(
    Total_Transactions=("Product", "count"),
    Total_Items_Bought=("Quantity", "sum"),
    Total_Spent=("Total_Value", "sum")
).reset_index()
display(customer_summary)

print("\n--- Product Summaries ---")
product_summary = df_organized.groupby("Product").agg(
    Total_Units_Sold=("Quantity", "sum"),
    Times_Purchased=("Customer", "count"),
    Total_Revenue=("Total_Value", "sum")
).reset_index()
display(product_summary)
```

```python
ordered_transactions = df_organized.sort_values(by="Total_Value", ascending=False).reset_index(drop=True)

print("======================================================")
print(" FINAL CONSOLIDATED TRANSACTION REPORT")
print("======================================================")
display(ordered_transactions)
```
